# Otras medidas de riesgo y rendimiento

```python
import pandas as pd; import numpy as np; import yfinance as yf
```

```python
tickers = ['WMT', 'COST']
```

```python
prices = yf.download(tickers, start='2020-01-01', end='2026-09-07')['Close']
prices
```

```python
rets = prices.pct_change().dropna()
rets
```

### Rendimiento esperado $E[R_i]$ (FL)

```python
rets.mean() * 252 * 100
```

**Si invierto en un activo a largo plazo, espero ganar en promedio $x$.**

### Media geométrica - $CAGR$ (Compound Annual Growth Rate) $R_G$ (BL)

Representa el rendimiento compuesto:

$$
R_G
=
\left(
\prod_{i=1}^{n}(1+r_i)
\right)^{\frac{1}{n}}
-1
$$

```python
CAGR = ((rets + 1).prod() ** (1 / len(rets))) - 1
CAGR * 252 * 100
```

**Si invierto mi capital y cada periodo reinvierto lo ganado, a largo plazo espero generar $x$.**

Nota: **$CAGR < E[R_i]$**, siempre y para todos los casos. Conforme la varianza aumenta, esta diferencia se nota aun más. 

El CAGR se considera si quieres considerar reinversiones en el capital y quieres ver tu <u>rendimiento real</u>, el promedio es solo para mirar a futuro y ver una proyección.

### Media armónica (BL)

Esta medida es muy útil cuando se trabaja con datos con volatilidad alta y con outliers, ya que mitiga su influencia en el resultado final:

$$
R_H
=
\frac{n}
{\displaystyle \sum_{i=1}^{n}\frac{1}{1+r_i}}
-1
$$

```python
MA = (len(rets) / ((1 / (rets + 1)).sum())) - 1
MA * 252 * 100
```

Este rendimiento armónico, al mitigar los outliers, siempre va a ser menor que el $CAGR$ y que la media.

Entre más varianza, más diferencia porcentual habrá en las medias.

## Medidas probabilísticas

### Rendimiento Esperado (FL)

Este rendimiento es probabilístico en el corto plazo.

```python
df = pd.DataFrame()
df['Escenario'] = ['Bajista', 'Neutral', 'Alcista']
df['Probabilidad'] = [0.25, 0.3, 0.45]
df['Rendimiento'] = [-10, 5, 20]

df
```

```python
E_R = (df['Probabilidad'] * df['Rendimiento']).sum()
E_R
```

**En base a las probabilidades de cada escenario económico, en el corto plazo espero ganar $x$.**

*Capital Market Expectations*

Es lo que se usa para calcular probabilidades.

### Varianza Esperada (FL)

```python
E_V = (df['Probabilidad'] * ((df['Rendimiento'] - E_R) ** 2)).sum()
E_V
```

### Volatilidad Esperada (FL)

```python
E_vol = np.sqrt(E_V)
E_vol
```
