Ahora, la elaboración del portafolio.

Importación de librerías

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

Definimos los tickers de los activos con los que vamos a construir nuestro portafolio

```python
ticker = ['ED', 'NUS', 'MDGL', 'TRMD']
```

Descargamos los precios de cierre en las fechas asignadas

```python
prices = yf.download(ticker,  start='2020-01-01', end='2026-08-31')['Close']
prices
```

Definimos los pesos de cada activo, en este caso será un portafolio equiponderado

```python
weights = np.ones(4) / 4
weights
```

Calculamos los rendimientos diarios de cada activo

```python
daily_rets= prices.pct_change().dropna()
daily_rets
```

### Rendimiento diario del portafolio

```python
daily_means = daily_rets.mean()
daily_means
```

```python
daily_ret_port = weights @ daily_means
daily_ret_port
```

### Rendimiento anual del portafolio

```python
annual_ret_port = daily_ret_port * 252
annual_ret_port
```

### Varianza diaria del portafolio

Primero requerimos la matriz de covarianzas

```python
cov_matrix = daily_rets.cov()
cov_matrix
```

```python
daily_risk_port = weights.T @ cov_matrix @ weights
daily_risk_port
```

### Varianza anual del portafolio

```python
annual_risk_port = daily_risk_port * 252
annual_risk_port
```

### Volatilidad diaria del portafolio

```python
daily_vol_port = np.sqrt(daily_risk_port)
daily_vol_port
```

### Volatilidad anual del portafolio


```python
annual_vol_port = daily_vol_port * np.sqrt(252)
annual_vol_port
```

### Matriz de correlación

```python
correlation_matrix = daily_rets.corr()
correlation_matrix
```

#### Rendimiento y Volatilidad anual de cada activo por separado y del portafolio

Esto nos servirá para poder ampliar nuestras conclusiones más adelante.

```python
# Calculamos la desviación estándar anual (Volatilidad)
annual_vol = daily_rets.std() * np.sqrt(252)

# Preparamos los datos para graficar
comparation_df = pd.DataFrame({
    'Retorno Anual (%)': daily_means * 252 * 100,
    'Volatilidad Anual (%)': annual_vol * 100
})

# Agregar las métricas del portafolio al DataFrame
comparation_df.loc['Portafolio'] = [annual_ret_port * 100, annual_vol_port * 100]

# Configuración de estilo
sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Gráfica de Rendimientos
sns.barplot(
    x=comparation_df.index, 
    y='Retorno Anual (%)', 
    data=comparation_df, 
    ax=axes[0], 
    palette='colorblind', 
    hue=comparation_df.index, 
    legend=False
)
axes[0].set_title('Rendimientos Anuales Esperados')
axes[0].axhline(0, color='black', linewidth=1)
for container in axes[0].containers:
    axes[0].bar_label(
        container, 
        fmt='%.2f%%', 
        label_type='edge'
    )

# Gráfica de Volatilidad 
sns.barplot(
    x=comparation_df.index, 
    y='Volatilidad Anual (%)', 
    data=comparation_df, 
    ax=axes[1], 
    palette='colorblind', 
    hue=comparation_df.index, 
    legend=False
)
axes[1].set_title('Volatilidad Anual')
for container in axes[1].containers:
    axes[1].bar_label(
        container, 
        fmt='%.2f%%', 
        label_type='edge'
    )

plt.show()
```

```python
# Gráfica de Riesgo vs Retorno 
plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=comparation_df, 
    x='Volatilidad Anual (%)', 
    y='Retorno Anual (%)', 
    s=200, 
    hue=comparation_df.index, 
    palette='colorblind'
)

# Mostrar portafolio con un marcador estrellado y más grande
plt.scatter(
    comparation_df['Volatilidad Anual (%)'].loc['Portafolio'], 
    comparation_df['Retorno Anual (%)'].loc['Portafolio'], 
    color='red', 
    s=300, 
    marker='*', 
    label='Portafolio'
)

# Añadir etiquetas a los puntos 
for i in range(comparation_df.shape[0]):
    plt.text(
        comparation_df['Volatilidad Anual (%)'].iloc[i]+0.5, 
        comparation_df['Retorno Anual (%)'].iloc[i],
        comparation_df.index[i], 
        fontsize=12, 
        weight='bold'
    )

plt.title('Análisis Comparativo: Riesgo vs. Retorno', fontsize=15)
plt.xlabel('Riesgo (Volatilidad %)')
plt.ylabel('Recompensa (Retorno Anual %)')
plt.grid(True)
plt.show()
```

### Conclusiones sobre la matriz de correlación y el riesgo no sistemático de cada activo

#### Matriz de Correlación

```python
# Graficar correlación
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix, 
    annot=True, 
    cmap='RdBu_r',
    vmin=-1, 
    vmax=1,
    center=0, 
    fmt='.5f', 
    linewidths=0.5   
)
plt.title('Matriz de Correlación')
plt.show()
```

Como se ve en la matriz de correlación, todas las correlaciones están bastante cercanas a cero.

El par con la mayor correlación positiva es MDGL-NUS, con aproximadamente 0.0473. Aunque es el valor más alto de la matriz, continúa siendo muy cercano a cero, por lo que no puede afirmarse que ambas acciones hayan tenido una tendencia constante a subir o bajar juntas. Por otro lado, ED-NUS presenta la correlación negativa de mayor magnitud, con aproximadamente -0.0251. Este resultado tampoco es suficientemente grande como para considerar que ED funcione como una protección frente a las caídas de NUS.

Las correlaciones entre TRMD y las demás acciones son prácticamente nulas. Esto sugiere que los cambios diarios en el precio de TORM estuvieron relacionados principalmente con factores diferentes de los que influyeron sobre Consolidated Edison, Nu Skin y Madrigal. Esta diferencia puede entenderse al observar sus actividades. TORM depende de las tarifas de transporte marítimo y de la operación de su flota, ED de servicios públicos regulados, NUS de las ventas de productos de belleza y bienestar y Madrigal de acontecimientos clínicos, regulatorios y comerciales relacionados con sus medicamentos.

La variedad de negocios ayuda a explicar la baja correlación, pero no la demuestra por sí sola. Dos empresas pueden pertenecer a sectores diferentes y, aun así, reaccionar de manera parecida ante una recesión, una crisis financiera, un aumento de tasas de interés o una caída general del mercado. Por ello, la matriz debe interpretarse como evidencia histórica de una baja relación lineal y no como una "garantía" de que las acciones continuarán comportándose de esa manera.

También es importante señalar que una correlación cercana a cero no significa que los activos sean independientes. La correlación resume el comportamiento promedio durante todo el periodo y puede ocultar días en los que las acciones se movieron en la misma dirección. Por ejemplo, el 16 de marzo de 2020 las cuatro posiciones registraron rendimientos negativos y el portafolio perdió aproximadamente 11.94% en una sola sesión. Esto demuestra que una baja correlación no impide que aparezcan pérdidas simultáneas durante periodos de tensión.

Por tanto, sí que existe un beneficio potencial de diversificación, debido a que los cuatro activos no se movieron de forma perfectamente sincronizada. Sin embargo, la matriz no permite concluir que el portafolio sea de bajo riesgo, que las acciones nunca caerán juntas o que las correlaciones se mantendrán constantes en el futuro.

#### Riesgo no sistemático de cada activo

El riesgo no sistemático es el relacionado con situaciones particulares de una empresa. Algunos ejemplos son una falla operativa, un producto que no logra venderse, un accidente, una decisión regulatoria específica o un resultado clínico inesperado. A diferencia de los acontecimientos generales del mercado, estos riesgos pueden reducirse al combinar empresas que dependan de actividades distintas, que es justo lo que se busca al diversificar un postafolio.

La correlación ayuda a observar si los efectos de esos riesgos particulares tienden a presentarse al mismo tiempo. Cuando las correlaciones son bajas, un problema en una empresa no necesariamente produce el mismo movimiento en las demás. Por ejemplo, un resultado clínico desfavorable para Madrigal no debería afectar directamente las tarifas reguladas de Consolidated Edison ni el número de viajes realizados por los buques de TORM. En ese sentido, la baja correlación puede disminuir la dependencia del portafolio respecto de un solo acontecimiento empresarial.

No obstante, la matriz de correlación no separa directamente el riesgo sistemático del no sistemático. La volatilidad calculada para cada acción incluye ambos, para medir estadísticamente el riesgo específico sería necesario comparar cada acción con un índice de mercado u otros factores y observar qué parte de sus movimientos no queda explicada por ellos.

### Conclusiones sobre las métricas de rendimiento y riesgo del portafolio, además de la diversificación del mismo 

#### Métricas del portafolio

```python
columns = ['Rendimiento diario (%)', 'Rendimiento anual (%)',  'Varianza diaria', 'Varianza anual', 'Volatilidad diaria (%)', 'Volatilidad anual (%)']
values = [daily_ret_port * 100, annual_ret_port * 100, daily_risk_port, annual_risk_port, daily_vol_port * 100, annual_vol_port * 100]

metrics_portafolio_df = pd.DataFrame([values], columns=columns, index=['Portafolio'])

metrics_portafolio_df.style \
    .hide(axis='index') \
    .set_caption('Rendimiento y Volatilidad del Portafolio') 
```

El portafolio se elaboró con pesos equitativos, es decir, 25% de peso para cada activo. Si analizamos el rendimiento y su volatilidad (anualizadas) por separado, vemos lo siguiente:

```python
# Métricas de activos individuales
metrics_individual_df = pd.DataFrame({
    'Rendimiento diario (%)': daily_means * 100,
    'Rendimiento anual (%)': daily_means * 252 * 100,
    'Varianza diaria': daily_rets.var(),
    'Varianza anual': daily_rets.var() * 252,
    'Volatilidad diaria (%)': daily_rets.std() * 100,
    'Volatilidad anual (%)': daily_rets.std() * np.sqrt(252) * 100
})

metrics_individual_df.style \
    .set_caption('Rendimiento y Volatilidad por Activo')
```

El portafolio presenta un rendimiento promedio diario aproximado de 0.1021% y uno anual de 25.72%. Esta cifra resume el promedio de los rendimientos diarios, pero no significa que el portafolio haya ganado exactamente 25.72% cada año, lo que nos dice esta cifra es que el en el largo plazo los rendimientos tienden a ese valor.

El resultado fue impulsado principalmente por Madrigal y TORM. ED realizó una aportación positiva más moderada, mientras que Nu Skin redujo el promedio, esto demuestra que el rendimiento total no proviene de manera equilibrada de las cuatro acciones, aunque todas tengan el mismo peso.

La volatilidad diaria del portafolio es aproximadamente 2.31% y su volatilidad anualizada es 36.67%. Esta última cifra representa la intensidad de las fluctuaciones históricas. La volatilidad del portafolio es menor que la de NUS, TRMD y MDGL, pero mayor que la de ED. Por consiguiente, la diversificación logró reducir el riesgo en comparación con tres de las cuatro inversiones individuales, pero no produjo una combinación menos volátil que todos sus componentes. 

Además, el rendimiento y la volatilidad deben analizarse juntos. Un rendimiento histórico elevado no es suficiente para calificar una inversión como favorable si fue necesario soportar oscilaciones muy grandes. En este caso, el rendimiento esperado del 25.72% estuvo acompañado de una volatilidad de 36.67% y de pérdidas importantes en ciertos momentos. Sin una referencia de mercado y una tasa libre de riesgo consistentes, no puede determinarse si la compensación por el riesgo fue superior a otras alternativas.

#### Diversificación del portafolio

Las métricas individuales muestran cuánto fluctuó y cuánto rindió cada activo por separado, la matriz de correlación muestra cómo se relacionaron esos movimientos, la volatilidad del portafolio reúne ambos elementos, es decir, depende tanto del riesgo individual de las acciones como de la forma en que se movieron entre sí. Por eso no basta con seleccionar acciones de baja correlación, ya que si una de ellas tiene una volatilidad extremadamente elevada, puede continuar dominando el riesgo del portafolio. Esto es precisamente lo que ocurre con MDGL, la correlación ayuda a reducir la transmisión de sus movimientos a las demás posiciones, pero su gran volatilidad hace que siga aportando aproximadamente dos terceras partes de la varianza.

El beneficio de la diversificación puede observarse mediante tres escenarios. Si los rendimientos de las cuatro acciones hubieran presentado una correlación positiva perfecta, la volatilidad equiponderada habría sido aproximadamente 63.63%. Con las correlaciones realmente observadas fue 36.67%. Si todas las correlaciones hubieran sido exactamente cero, habría resultado cercana a 36.13%. La diferencia entre 63.63% y 36.67% muestra que combinar activos que no se mueven juntos reduce considerablemente el riesgo frente a una situación de sincronización perfecta, sin embargo, la cercanía entre 36.67% y 36.13% señala que casi todo el riesgo restante proviene de las volatilidades individuales, especialmente de Madrigal (lo cual se entiende debido al caso extremo que presentó en el periodo de estudio).

Esto también permite distinguir entre repartir el capital y repartir el riesgo, cada activo recibe 25% del dinero, pero sus aportaciones a la varianza son muy diferentes. ED aporta aproximadamente 2.62%, NUS un 15.24%, TRMD un 15.68% y MDGL un 66.46%. En consecuencia, el portafolio está equiponderado en términos monetarios, pero no en términos de riesgo.

En conclusión, se podría decir que el portafolio presenta una diversificación favorable porque reúne empresas de cuatro actividades económicas diferentes y porque sus rendimientos históricos tuvieron correlaciones pequeñas. Esto disminuye la dependencia de una sola empresa y evita que todos los riesgos particulares se transmitan automáticamente a la combinación, sin embargo, la diversificación no está del todo "bien" hecha. Sólo existen cuatro emisores, cada uno representa una parte importante del capital y todas las posiciones pertenecen a renta variable, por lo que no hay diversificación entre diferentes clases de activos y permanece una fuerte concentración del riesgo en Madrigal. 

La prueba en la que se retira el rendimiento de Madrigal del 19 de diciembre de 2022 refuerza la idea de que minimizar la correlación no siempre minimiza la volatilidad, porque si cuando se omite esa observación, la media anualizada y la volatilidad disminuyen, mientras que las correlaciones aumentan bastante, mostrando que un solo acontecimiento extraordinario influye considerablemente en varias métricas. La observación no debe eliminarse del resultado principal, puesto que es un dato verídico que sucedió, pero sí invita a reconocer que la relación entre los activos también cambia dependiendo del periodo y de eventos extremos.
