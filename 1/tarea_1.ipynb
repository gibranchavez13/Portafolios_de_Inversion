### **Precios de cierre ajustado de `PRGO`, `PLTR` y `BRK-B` del 2018-01-01 al 2026-08-24**

Importación de librerías

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

Tickers de los activos a trabajar

```python
ticker = ['PRGO', 'PLTR', 'BRK-B']
```

**Descarga de precios de cierre**

```python
prices = yf.download(ticker,  start='2018-01-01', end='2026-08-24')['Close']
prices
```

Distribuciones de los precios de cierre

```python
colors = ['blue', 'lightblue', 'darkblue']
```

```python
for i, j in zip(ticker, colors):
    plt.figure(figsize=(12, 6))
    series = prices[i].dropna()
    plt.hist(series, bins=100, color=j)
    plt.title(f'Histograma de precios de {i}')
    plt.show()
```

Como se ve en las gráficas y como se vio en clase, los precios nunca siguen una distribución. No se puede usar el precio para modelar el comportamiento futuro de una acción sin hacer transformaciones.

Por eso se usan los rendimientos.

------------

**Cálculo de rendimeintos diarios**

```python
daily_rets = (prices/prices.shift() - 1).dropna()
daily_rets
```

Nota: Como PLTR empezó a cotizar desde 2020-10-01, el análisis de los tres activos realmente comienza desde esa fecha.

------------

**Distribuciones de los rendimientos diarios**

```python
for i, j in zip(ticker, colors):
    plt.figure(figsize=(12, 6))
    plt.hist(daily_rets[i], bins=100, color=j)
    plt.title(f'Histograma de rendimientos diarios de {i}')
    plt.show()
```

**Rendimiento anual promedio**

```python
daily_means = daily_rets.mean()
annual_means = daily_means * 252
annual_means
```

```python
rendimientos_df = (
    annual_means.mul(100).rename('Rendimiento promedio anualizado (%)').reset_index()
)

rendimientos_df.style \
    .hide(axis='index') \
    .format({'Rendimiento promedio anualizado (%)': '{:.4f}%'}) \
    .set_caption('Rendimientos promedio anualizados por activo')
```

En relación con los activos, los 3 muestran situaciones que son diferentes en sus rendimientos, siendo que BRK-B muestra un rendimiento positivo y algo realista, PLTR con un rendimiento bastante alto, aunque a simple vista uno se pregunta qué riesgos hay en relación a ese rendimiento tan alto, y con PRGO, vemos un rendimiento negativo, nos dice que está perdiendo valor, para unos es una oportunidad de compra, para otros una señal de salida.

PLTR presentó el mayor rendimiento anual promedio, con aproximadamente 74.19%, seguido de BRK-B con 15.89%. En contraste, PRGO registró un rendimiento promedio negativo de -9.24%, indicando una pérdida de valor promedio durante el periodo analizado. Sin embargo, estos resultados no son suficientes para determinar cuál activo es la mejor inversión, ya que un mayor rendimiento puede estar acompañado de un mayor nivel de riesgo. 

---------

**Riesgo anual**

```python
daily_var = daily_rets.var()
annual_var = daily_var * 252
```

```python
var_df = (
    annual_var.rename('Riesgo promedio anualizado').reset_index()
)

var_df.style \
    .hide(axis='index') \
    .format({'Riesgo promedio anualizado (%)': '{:.4f}%'}) \
    .set_caption('Riesgo promedio anualizado por activo')
```

**Volatilidad anual**

```python
daily_std = daily_rets.std()
annual_std = daily_std * np.sqrt(252)
```

```python
vol_df = (
    annual_std.mul(100).rename('Volatilidad promedio anualizada (%)').reset_index()
)

vol_df.style \
    .hide(axis='index') \
    .format({'Volatilidad promedio anualizada (%)': '{:.4f}%'}) \
    .set_caption('Volatilidad (desviación estándar) promedio anualizada por activo')
```

**Coeficiente de variación**

```python
CV = annual_std / annual_means
```

```python
cv_df = (
    CV.rename('CV').reset_index()
)

cv_df.style \
    .hide(axis='index') \
    .set_caption('Coeficiente de variación por activo')
```

BRK-B presenta un rendimiento promedio anual de 15.89%, acompañado de una volatilidad de 17.23%. De los 3 activos, es el que tiene una menor volatilidad, y su coeficiente de variación es de 1.0845, que quiere decir que por cada 1% de rendimiento promedio, BRK-B presenta alrededor de 1.08% de volatilidad. Por lo tanto, BRK-B tiene un comportamiento relativamente más estable. No ofrece el grandísimo rendimiento como PLTR, pero tampoco parece tener fluctuaciones tan grandes. 

Después está PLTR, que ofrece un rendimiento altísimo del 74.19%, aunque también tiene una volatilidad del 70.35%, lo cual lo convierte en el activo más riesgoso de los 3. Sin embargo, tiene un coeficiente de variación de 0.9482, que son aproximadamente 0.95 unidades de volatilidad por cada unidad de rendimiemto. Osea que PLTR es más riesgoso, pero tiene una mayor cantidad de rendimiento por unidad de riesgo asumida.

En el caso del PRGO, no hay mucho que decir. Con un rendimiento promedio de -9.24%, una volatilidad de 37.54%, es decir, es un activo considerablemente riesgoso y aparte genera pérdidas en el largo plazo. Su coeficiente de variación es de -4.06, esto a causa de su rendimiento negativo. Entonces esta última métrica podría volverse algo engañosa al comparar la relación rendimiento-riesgo. 

Hasta ahora con estos datos, considerando solo el riesgo:

$$
BRK-B < PRGO < PLTR
$$

Lo cual pone a BRK-B como el activo menos riesgoso, pero si ahora solo tomamos en cuenta el rendimiento:

$$
PLTR > BRK-B > PRGO
$$

Pero tomando en cuenta la relación entre estod dos, `PLTR` sería la mejor opción, ya que como está explicado anteriormente, tiene un $CV = 0.9482$, lo cual es el mejor de los 3, ya que es el que te permite asumir menor volatilidad por cada unidad de rendimiento.

---------

**Asimetría**

```python
sk = daily_rets.skew()
```

```python
sk_df = (
    sk.rename('Skew').reset_index()
)

sk_df.style \
    .hide(axis='index') \
    .set_caption('Coeficiente de asimetría por activo') 
```

Al agregar el coeficiente de asimetría al análisis, ahora se puede conocer hacia qué lado tienden a presentarse los eventos extremos. 

Para BRK-B, hay un asimetría de 0.0810, lo cual está bastante cerca del 0, esto quiere decir que su distribución es aproximadamente simétrica y que no se observa una tendencia marcada hacia grandes movimientos positivos o negativos. Sus rendimientos extremos se encuentran relativamente equilibrados entre ambos lados de la distribución (como se puede ver en la distribución de sus rendimientos). Esto concuerda con el rendimiento y la volatilidad antes observada, ya que hay menor varianza.

Después, PLTR tiene una asimetría de 1.1420, lo cual indica una asimetría positiva algo considerable, con una cola más larga hacia la derecha. Este dato complementa el análisis de la volatilidad, ya que, esta es de 70.35%, pero ahora se sabe que una parte de ese resultado es a causa de movimientos extremos hacia el lado positivo de la distribución (como se ve en la distribución de sus rendimientos). Sigue siendo un activo muy volátil y riesgoso, pero los extremos positivos han sido más pronunciados que los negativos.

Por último, PRGO muestra una asimetría de -0.0986, que está bastante cerca de 0, por lo que es una asimetría negativa ligera. Aun asi, la inclinación hacia rendimientos negativos existe, por lo que si distribución tiene una cola ligeramente mayor hacia las pérdidas, y tiene sentido, si vemos su rendimeinto anual promedio, es negativo.

En cuanto al mejor activo de los 3, `PLTR` presenta el perfil más atractivo en términos de rendimiento-riesgo. Aunque tiene la volatilidad más alta, también tiene un rendimiento promedio muy superior a los demás, también un menor coeficiente de variación que BRK-B y una asimetría positiva considerable de 1.142, lo que indica una mayor presencia de rendimientos extremos positivos. Por ello, al incorporar el coeficiente de asimetría no cambia la conclusión anterior, sino que la refuerza ligeramente.

Aunque si solo se priorizara la estabilidad, BRK-B sería la mejor alternativa. Por lo que aquí entraría a tema la adversión al riesgo de cada persona.

```python
final_df = (
    rendimientos_df
    .merge(var_df, on='Ticker', how='inner')
    .merge(vol_df, on='Ticker', how='inner')
    .merge(cv_df, on='Ticker', how='inner')
    .merge(sk_df, on='Ticker', how='inner')
)

final_df.style \
    .hide(axis='index') \
    .set_caption('Tabla de métricas')
```
